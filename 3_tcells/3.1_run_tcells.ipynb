## Train the model and predict on the new dataset

Before following steps, please make sure the Cano-Gamez dataset is downloaded and preprocessed properly in the `../datasets/canogamez` directory.

If you want to try other settings, please modify the `config_tcells.yaml` file.

```python
pp = Pipeline(config_path='config_tcells.yaml')
pp.fit()
pp.save_best_model()
adata_pred = pp.predict(num_results=5)
os.makedirs(os.path.join(pp.args.trainer.output_dir, 'anndata'), exist_ok=True)
for key in adata_pred.keys():
    adata_pred[key].write(os.path.join(
        pp.args.trainer.output_dir, 
        'anndata', 
        f'{key}.h5ad'
    ))
```
