```python
import os
os.environ['CUDA_VISIBLE_DEVICES'] = '0'
import pandas as pd
import driver_genes as dg
dg.set_verbosity(False)

from driver_genes.pipelines import Pipeline
from driver_genes.myutils import render_template
```

## load job info and config template

The config template is a yaml file that contains the placeholder of parameters for the experiment.

changable parameters: 
- In job info: job_name, split_version, adata_file, split_file, layer_name, subset_name
- Additional: output_dir


```python
job_info_df = pd.read_csv('job_info_benchmark.csv')
config_template_path = "config_template.yaml"

# specify other experiment parameters
output_dir = '../results/benchmarking'
```

```python
for info_dict in job_info_df.to_dict(orient='records'):
    print(info_dict)
    config_dict = render_template(config_template_path, **info_dict, output_dir=output_dir)
    pipeline = Pipeline(config_dict=config_dict)
    pipeline.fit()
    pipeline.save_best_model()
    metrics_df, adata_pred = pipeline.evaluate(num_results=5, concat=True, agg=True)
    metrics_df.to_csv(os.path.join(pipeline.args.trainer.output_dir, 'test_epoch_metrics.csv'), index=False)
    os.makedirs(os.path.join(pipeline.args.trainer.output_dir, 'anndata'), exist_ok=True)
    for key in adata_pred.keys():
        adata_pred[key].write(os.path.join(
            pipeline.args.trainer.output_dir, 
            'anndata', 
            f'{key}.h5ad'
        ))
    break
```

## obtain predicted probabilities for each cell

Here we only predict Schmidt dataset and Norman dataset for reproducing the figures in paper.

```python
from driver_genes.bmutils import label_metrics, double_relpos_distribution
```

```python
expr_list = [
    "../results/benchmarking/schmidt/resting", 
    "../results/benchmarking/schmidt/stimulated", 
    "../results/benchmarking/norman/combo_seensinglesbetter"
]
```

```python
for expr_prefix in expr_list:
    expr_dir = f'{expr_prefix}-seed42'
    os.makedirs(f'{expr_dir}/results', exist_ok=True)
    # load fitted model to predict and not aggregating results
    pp = Pipeline(config_path=os.path.join(expr_dir, 'config.yaml'), fitted=True)
    metrics_df, test_pred_adatas_list = pp.evaluate(num_results=5, seed=42, concat=False)
    results_df_dict = {}
    for i, test_pred_adatas in enumerate(test_pred_adatas_list):
        for ad_name, ad in test_pred_adatas.items():
            ad = ad[ad.obs['perturbation'] != 'control']
            results_df_dict[ad_name] = results_df_dict.get(ad_name, [])
            results_df_dict[ad_name].append(ad.obsm['proba'].assign(perturbation=ad.obs['perturbation'], rep=i))
    for ad_name, results_df_list in results_df_dict.items():
        results_df = pd.concat(results_df_list, axis=0, ignore_index=True)
        results_df.to_parquet(os.path.join(expr_dir, f'results/{ad_name}_results.parquet'), index=False)
```

```python
task_dict = {
    # (Dataset, Test on)
    ('Schmidt', 'Resting'): {
        'Ours': '../results/benchmarking/schmidt/stimulated-seed42/results/preprocessed_Resting_results.parquet',
        # path to results from other methods
    },
    ('Schmidt', 'Re-stimulated'): {
        'Ours': '../results/benchmarking/schmidt/resting-seed42/results/preprocessed_Re-stimulated_results.parquet',
        # path to results from other methods
    },
    ('Norman', 'Single'): {
        'Ours': '../results/benchmarking/norman/combo_seensinglesbetter-seed42/results/preprocessed_single_results.parquet',
        # path to results from other methods
    },
    ('Norman', 'Combo'): {
        'Ours': '../results/benchmarking/norman/combo_seensinglesbetter-seed42/results/preprocessed_combo_results.parquet',
        # path to results from other methods
    },
}
```

```python
results_dict = {}
for (dataset, cell_type), res_path_dict in task_dict.items():
    res_list = []
    for method, res_path in res_path_dict.items():
        results_df = pd.read_parquet(res_path)
        res_list.append(results_df.assign(method=method))
    results_dict[(dataset, cell_type)] = pd.concat(res_list)
pd.to_pickle(results_dict, 'data/results_dict_runtime.pkl')
```

```python
def group_apply(subres_df):
    unness_cols = list(set(subres_df.columns) & set(['perturbation', 'rep', 'method']))
    score_df = subres_df.drop(columns=unness_cols)
    perturbations = subres_df['perturbation']
    metrics = label_metrics(score_df, perturbations)
    return metrics
```

### Schmidt dataset

```python
results_df = pd.concat([
    results_dict[('Schmidt', 'Resting')].assign(cell_type='Resting'),
    results_dict[('Schmidt', 'Re-stimulated')].assign(cell_type='Re-stimulated'),
])
label_metrics_df = (
    results_df
    .groupby(['cell_type', 'method', 'rep'])
    .apply(group_apply, include_groups=False)
    .reset_index()
    .groupby(['cell_type', 'method', 'perturbation'])
    .agg(['mean', 'std'])
    .reset_index()
    .drop(columns=['rep', 'level_3'])
)
```

```python
other_metrics_df = pd.read_pickle('data/schmidt_label_metrics_df.pkl')
# filter out the metrics of our model from the previous runnings
other_metrics_df = other_metrics_df.query('Method != "Ours"')
all_metrics_df = pd.concat([
    other_metrics_df,
    label_metrics_df
])
label_metrics_df.to_pickle('data/schmidt_label_metrics_df.pkl')
```

### Norman dataset

```python
def group_apply(subres_df):
    unness_cols = list(set(subres_df.columns) & set(['perturbation', 'rep', 'method']))
    score_df = subres_df.drop(columns=unness_cols)
    perturbations = subres_df['perturbation']
    relpos = double_relpos_distribution(score_df, perturbations)
    return relpos
```

```python
results_df = results_dict[('Norman', 'Combo')]
relpos_df = (
    results_df
    .groupby(['method'])
    .apply(group_apply, include_groups=False)
    .reset_index()
    .drop(columns=['level_1'])
)
```

```python
other_relpos_df = pd.read_pickle('data/norman_relpos_df.pkl')
# filter out the relpos of our model from the previous runnings
other_relpos_df = other_relpos_df.query('Method != "Ours"')
all_relpos_df = pd.concat([
    other_relpos_df,
    relpos_df
])
all_relpos_df.to_pickle('data/norman_relpos_df.pkl')
```
